```python
#strip 11 to 20
import os
import numpy as np
import pandas as pd
import cv2
from pathlib import Path
from tqdm import tqdm

# ==========================
# CONFIG
# ==========================
STRIPS_ROOT = r"D:\OHRC_Project\OHRC"   # folder where your strip folders (1,2,3,...) are
OUTPUT_ROOT = r"D:\OHRC_Project\tiles"  # output tiles folder
TILE_SIZE = 640
BLACK_THRESHOLD = 5
START_STRIP = 11   # first strip to process
END_STRIP   = 20   # last strip to process (inclusive)

# ==========================
# FUNCTION
# ==========================
def process_strip(strip_path, out_dir, tile_size=640):
    img_file = [f for f in os.listdir(strip_path) if f.endswith(".img")][0]
    csv_file = [f for f in os.listdir(strip_path) if f.endswith(".csv")][0]

    IMG_FILE = os.path.join(strip_path, img_file)
    CSV_FILE = os.path.join(strip_path, csv_file)

    df = pd.read_csv(CSV_FILE)
    width = int(df["Pixel"].max() + 1)
    height = int(df["Scan"].max() + 1)

    mm = np.memmap(IMG_FILE, dtype=np.uint8, mode="r", shape=(height, width))

    Path(out_dir).mkdir(parents=True, exist_ok=True)
    clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8,8))
    log_rows, tile_id = [], 0

    print(f"Processing {Path(strip_path).name}: size {height}x{width}")

    for row in tqdm(range(0, height, tile_size)):
        for col in range(0, width, tile_size):
            tile = mm[row:row+tile_size, col:col+tile_size]
            if tile.shape != (tile_size, tile_size): 
                continue
            if tile.mean() < BLACK_THRESHOLD: 
                continue

            raw_name = f"tile_{tile_id}_raw.png"
            clahe_name = f"tile_{tile_id}_clahe.png"
            cv2.imwrite(os.path.join(out_dir, raw_name), tile)
            cv2.imwrite(os.path.join(out_dir, clahe_name), clahe.apply(tile))

            # center pixel → lat/lon
            center_row, center_col = row + tile_size//2, col + tile_size//2
            match = df[(df["Pixel"] == center_col) & (df["Scan"] == center_row)]
            lat, lon = (match["Latitude"].values[0], match["Longitude"].values[0]) if not match.empty else (None, None)

            log_rows.append([raw_name, row, col, lat, lon])
            log_rows.append([clahe_name, row, col, lat, lon])
            tile_id += 1

    log_df = pd.DataFrame(log_rows, columns=["filename","row","col","lat","lon"])
    log_df.to_csv(os.path.join(out_dir, "tiles_log.csv"), index=False)

    print(f"✅ {Path(strip_path).name}: Saved {tile_id} tiles in {out_dir}")

# ==========================
# MAIN LOOP
# ==========================
for strip_num in range(START_STRIP, END_STRIP + 1):
    strip_path = os.path.join(STRIPS_ROOT, str(strip_num))
    if not os.path.isdir(strip_path):
        print(f"⚠️ Strip {strip_num} not found, skipping...")
        continue

    out_dir = os.path.join(OUTPUT_ROOT, f"strip{strip_num}")
    process_strip(strip_path, out_dir, TILE_SIZE)
```
